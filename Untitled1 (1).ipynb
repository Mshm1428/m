```python
# generate_and_prepare.py
"""
Improved generator: ensures each article is STRICTLY >1200 words (if OpenAI missing),
removes internal repetition, avoids similarity vs existing articles,
generates pin image (Pillow textbbox-safe) and saves HTML into publish_ready_cleaned/.
"""

import os, re, time, json, random
from pathlib import Path
from hashlib import md5
from difflib import SequenceMatcher

# OpenAI new client (optional)
try:
    from openai import OpenAI
    OPENAI_AVAILABLE = True
except Exception:
    OPENAI_AVAILABLE = False

# Pillow for image generation
try:
    from PIL import Image, ImageDraw, ImageFont
except Exception:
    Image = ImageDraw = ImageFont = None

# load .env if present (optional)
try:
    from dotenv import load_dotenv
    load_dotenv()
except Exception:
    pass

ROOT = Path.cwd()
TRENDS = ROOT / "trends_list.txt"
OUT_DIR = ROOT / "publish_ready_cleaned"
PIN_DIR = ROOT / "pin_images"
PUBLISHED_DIR = ROOT / "published"
OUT_DIR.mkdir(exist_ok=True)
PIN_DIR.mkdir(exist_ok=True)
PUBLISHED_DIR.mkdir(exist_ok=True)

MIN_WORDS = 1200   # must be strictly greater
SIMILARITY_THRESHOLD = 0.86
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
MODEL = os.getenv("OPENAI_MODEL", "gpt-3.5-turbo")
FONT_CANDIDATES = ["DejaVuSans-Bold.ttf", "arial.ttf"]

# ----------------- utilities -----------------
def word_count(html_text):
    plain = re.sub(r"<[^>]+>", " ", html_text)
    words = [w for w in re.split(r"\s+", plain) if w.strip()]
    return len(words)

def safe_name(s):
    s = s.strip()
    name = "".join(ch if ch.isalnum() or ch in "-_" else "_" for ch in s)[:90]
    return name or "article"

def similar(a,b):
    return SequenceMatcher(None, a, b).ratio()

def read_existing_texts():
    texts=[]
    for p in list(OUT_DIR.glob("*.html")) + list(PUBLISHED_DIR.glob("*.html")):
        try:
            texts.append(p.read_text(encoding="utf-8"))
        except Exception:
            pass
    return texts

# ----------------- image generator (Pillow textbbox-safe) -----------------
def generate_pin_image(title, out_path: Path):
    if Image is None:
        return False, "Pillow not installed"
    W,H = 1000,1500
    img = Image.new("RGB",(W,H),(255,255,255))
    draw = ImageDraw.Draw(img)
    font = None
    for f in FONT_CANDIDATES:
        try:
            font = ImageFont.truetype(f,56); break
        except Exception:
            continue
    if font is None:
        font = ImageFont.load_default()
    words = title.split()
    lines=[]; cur=""
    for w in words:
        cand = (cur+" "+w).strip() if cur else w
        bbox = draw.textbbox((0,0), cand, font=font)
        width = bbox[2]-bbox[0]
        if width < W-120:
            cur = cand
        else:
            if cur: lines.append(cur)
            cur = w
    if cur: lines.append(cur)
    heights=[]; total_h=0
    for ln in lines:
        b = draw.textbbox((0,0), ln, font=font)
        h = b[3]-b[1]; heights.append(h); total_h += h
    total_h += max(0,(len(lines)-1))*12
    y = (H - total_h)//2
    for ln,h in zip(lines, heights):
        b = draw.textbbox((0,0), ln, font=font)
        w = b[2]-b[0]; x = (W - w)//2
        draw.text((x,y), ln, fill=(20,20,20), font=font)
        y += h + 12
    out_path.parent.mkdir(parents=True, exist_ok=True)
    img.save(out_path, format="PNG", optimize=True)
    return True, out_path

# remove consecutive duplicate sentences & small repeated phrases
def remove_consecutive_duplicate_sentences(text):
    parts = re.split(r'(?<=[\.\!\?])\s+', text)
    cleaned=[]
    for p in parts:
        p = p.strip()
        if not p: continue
        if cleaned and p == cleaned[-1]:
            continue
        cleaned.append(p)
    return " ".join(cleaned)

# paragraph dedupe (exact + near-exact)
def paragraph_dedupe(text):
    paras = [p.strip() for p in re.split(r'\n{2,}|\r\n{2,}', text) if p.strip()]
    out=[]; seen=set()
    for p in paras:
        key = re.sub(r'\s+',' ', p)
        if key in seen: continue
        dup=False
        for existed in out:
            if similar(existed, p) > 0.98:
                dup=True; break
        if dup: continue
        seen.add(key); out.append(p)
    return "\n\n".join(out)

# ------------- deterministic expansion helpers -------------
TRANSITIONS = ["To begin with,", "Firstly,", "Moreover,", "In addition,", "Notably,", "Consequently,", "Importantly,", "For example,"]
TEMPLATES = [
    "{topic} has attracted growing interest across professionals and researchers. {extra}",
    "A practical way to understand {topic} is to consider concrete examples and outcomes. {extra}",
    "Industry practitioners often notice recurring patterns when it comes to {topic}. {extra}",
    "From a historical perspective, {topic} evolved due to practical needs and technological changes. {extra}",
    "When people apply {topic} in real contexts, several practical considerations emerge. {extra}",
    "A common misconception about {topic} is that it's only theoretical; in practice {extra}",
    "Stakeholders usually evaluate {topic} by focusing on benefits, constraints, and tradeoffs. {extra}",
]

EXTRA_SENTENCES = [
    "This paragraph explains the main idea and then gives a realistic example that readers can relate to.",
    "The following lines describe concrete steps and best practices that make the concept actionable.",
    "Understanding this helps readers make better decisions and avoid common mistakes in real situations.",
    "These points are based on common patterns observed by experienced practitioners.",
    "This perspective clarifies how the topic impacts day-to-day workflows and longer-term plans."
]

FAQ_QA = [
    ("What is the main idea?", "The main idea is to focus on principles, practical examples, and careful application in real contexts."),
    ("How can I start applying this?", "Begin with small experiments, measure outcomes, iterate, and learn from practical feedback."),
    ("What are common pitfalls?", "Overgeneralizing from a single example and ignoring context-specific constraints are typical pitfalls."),
]

def make_paragraph(topic, seed_idx, variant_idx):
    tpl = TEMPLATES[seed_idx % len(TEMPLATES)]
    extra = EXTRA_SENTENCES[variant_idx % len(EXTRA_SENTENCES)]
    trans = TRANSITIONS[(seed_idx + variant_idx) % len(TRANSITIONS)]
    return f"{trans} {tpl.format(topic=topic, extra=extra)}"

def generate_long_article(topic, target_words=1400):
    seed_int = int(md5(topic.encode('utf-8')).hexdigest()[:8], 16)
    random.seed(seed_int)
    paras = []
    paras.append(f"{topic} is an important subject gaining attention from many audiences. This article gives a clear, human explanation and practical guidance.")
    section_count = 10
    for s in range(section_count):
        for v in range(2 + (s % 3)):
            p = make_paragraph(topic, s, v)
            p = p + (" " + random.choice(EXTRA_SENTENCES))
            paras.append(p)
        if s % 3 == 0:
            paras.append(f"Subtopic {s+1}: A focused look at a practical angle of {topic}. It highlights consequences and concrete steps.")
    for i in range(6):
        paras.append(f"For example, consider scenario {i+1} where {topic} is applied: this leads to measurable outcomes and lessons learned.")
    for i in range(6):
        paras.append(f"Practical tip {i+1}: Start small, measure, adapt, and document results when working with {topic}.")
    for q,a in FAQ_QA:
        paras.append(f"Q: {q}\nA: {a}")
    paras.append(f"In conclusion, {topic} benefits from careful analysis, testing, and a focus on practical application. Further reading and references help deepen understanding.")
    paras.append("Sources: https://example.com/article1, https://example.com/article2")
    unique = []
    seen = set()
    for p in paras:
        key = re.sub(r'\s+',' ', p.strip())
        if key in seen:
            continue
        near_dup = False
        for up in unique[-6:]:
            if similar(up, p) > 0.95:
                near_dup = True; break
        if near_dup: continue
        seen.add(key); unique.append(p)
    html = "<h1>{}</h1>\n".format(topic)
    for p in unique:
        if p.startswith("Q: "):
            parts = p.split("\n")
            for part in parts:
                html += f"<p>{part.strip()}</p>\n"
        else:
            html += f"<p>{p.strip()}</p>\n"
    if word_count(html) <= target_words:
        attempt = 0
        while word_count(html) <= target_words and attempt < 60:
            extra_para = random.choice(EXTRA_SENTENCES) + " " + f"Additional insight about {topic} that helps understanding and application."
            if all(similar(extra_para, up) < 0.95 for up in unique):
                html += f"<p>{extra_para}</p>\n"
                unique.append(extra_para)
            attempt += 1
    return html

# ----------------- OpenAI wrapper (optional) -----------------
def openai_generate_article(topic, client, tries=2):
    system = ("You are a professional, native-level English content writer. Produce a human-sounding article "
              "with headings, examples, tips, and a Sources list. Avoid repeating paragraphs.")
    user = (f"Write a high-quality article about: {topic}\n"
            f"- The article MUST be MORE THAN {MIN_WORDS} words (strictly greater).\n"
            "- Avoid repeating whole paragraphs. Use varied phrasing and real-looking examples.\n"
            "- End with a 'Sources:' list of at least 2 links (placeholders ok).")
    for _ in range(tries):
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                messages=[{"role":"system","content":system},{"role":"user","content":user}],
                max_tokens=2500,
                temperature=0.7,
            )
            content = resp.choices[0].message.content.strip()
            if len(re.split(r'\s+', re.sub(r'<[^>]+>',' ',content))) > MIN_WORDS:
                return content
        except Exception as e:
            last_err = e
    raise RuntimeError(f"OpenAI generation failed or produced short output: {last_err if 'last_err' in locals() else 'no details'}")

# ----------------- main flow -----------------
def main():
    if not TRENDS.exists():
        print("trends_list.txt missing."); return
    topics = [t.strip() for t in TRENDS.read_text(encoding="utf-8").splitlines() if t.strip()]
    existing = read_existing_texts()
    client = None
    if OPENAI_AVAILABLE and OPENAI_API_KEY:
        client = OpenAI(api_key=OPENAI_API_KEY)
        print("OpenAI client available.")
    else:
        print("OpenAI client not available -> deterministic fallback.")
    generated = 0
    for topic in topics:
        fname = safe_name(topic) + ".html"
        outp = OUT_DIR / fname
        if outp.exists():
            print("skip exists:", fname); continue
        # generate: prefer OpenAI when available
        html_content = ""
        if client:
            try:
                html_content = openai_generate_article(topic, client)
            except Exception as e:
                print("OpenAI failed:", e, "-> fallback deterministic")
                html_content = generate_long_article(topic, target_words=MIN_WORDS+50)
        else:
            html_content = generate_long_article(topic, target_words=MIN_WORDS+50)
        # final cleaning: remove consecutive duplicate sentences
        html_content = re.sub(r'(?<=[\.\!\?])\s{2,}', ' ', html_content)
        if not (word_count(html_content) > MIN_WORDS):
            html_content += generate_long_article(topic, target_words=MIN_WORDS+50)
        too_similar = False
        for ex in existing:
            if similar(ex, html_content) > SIMILARITY_THRESHOLD:
                print("skip similar to existing:", topic); too_similar=True; break
        if too_similar: continue
        outp.write_text(html_content, encoding="utf-8")
        img_name = safe_name(topic) + ".png"
        img_path = PIN_DIR / img_name
        if not img_path.exists() or img_path.stat().st_size==0:
            ok,msg = generate_pin_image(topic, img_path)
            print("pin:", img_path.name if ok else "error "+str(msg))
        print("Saved:", outp.name, "| words:", word_count(html_content))
        generated += 1
    print("Done. generated:", generated)

if __name__ == "__main__":
    main()
```
