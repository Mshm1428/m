```python
#!/usr/bin/env python3
# fetch_trends_all.py
# يجلب "ترندات" من Hacker News + Reddit (يوميًا) ويحفظها في trends_list.txt
# متين: يحاول، يتخطى الأخطاء، يزيل التكرار، يطبع تقرير.

from pathlib import Path
import requests, time, re, os, sys, logging
from typing import List, Set

# --------- إعدادات قابلة للتعديل ----------
OUT_FILE = Path("trends_list.txt")
TOP_N = 40                # عدد العناصر النهائية التي نريد حفظها
HN_MAX = 80               # عدد الـ top IDs من Hacker News نقرأها
REDDIT_LIMIT = 120        # limit for reddit results fetch
TIMEOUT = 10              # ثانية للطلبات HTTP
SLEEP_BETWEEN = 0.12      # تأخير بسيط بين طلبات HN لتخفيف الضغط
HEADERS = {"User-Agent": "trend-fetcher-script/1.0 (+https://example.com)"}
# -----------------------------------------

logging.basicConfig(level=logging.INFO, format="%(message)s")

SAMPLE_FALLBACK = [
    "Artificial Intelligence",
    "Healthy Eating",
    "Sustainable Energy",
    "Remote Work Trends",
    "Mental Health Tips",
    "Smart Home Gadgets",
    "Personal Finance Tips",
    "Home Workouts",
    "Climate Action News",
    "Electric Vehicles"
]

# تنظيف العنوان: احتفظ بحروف الإنجليزية والعربية والأرقام وبعض علامات الترقيم البسيطة
def clean_text(s: str) -> str:
    if not s:
        return ""
    t = str(s).strip()
    # allow Arabic, Latin letters, numbers, basic punctuation (-:,.())
    t = re.sub(r"[^A-Za-z\u0600-\u06FF0-9\s\-\:\,\(\)\.']", " ", t)
    t = re.sub(r"\s{2,}", " ", t).strip()
    # Capitalize first letter for English looking words (improve appearance)
    def cap_word(w):
        return w.capitalize() if re.search(r"[A-Za-z]", w) else w
    return " ".join(cap_word(x) for x in t.split())

# ===== Hacker News =====
def fetch_hackernews(max_ids: int = HN_MAX) -> List[str]:
    base = "https://hacker-news.firebaseio.com/v0"
    results = []
    try:
        r = requests.get(f"{base}/topstories.json", headers=HEADERS, timeout=TIMEOUT)
        r.raise_for_status()
        ids = r.json()
        if not isinstance(ids, list):
            logging.warning("HN: unexpected topstories format.")
            return []
        count = 0
        for id_ in ids[:max_ids]:
            try:
                it = requests.get(f"{base}/item/{id_}.json", headers=HEADERS, timeout=TIMEOUT)
                it.raise_for_status()
                data = it.json() or {}
                title = data.get("title") or ""
                title = clean_text(title)
                if title:
                    results.append(title)
                    count += 1
                if count >= TOP_N:
                    break
            except Exception as e:
                # لا نوقف الكل بسبب خطأ في عنصر واحد
                logging.debug(f"HN item error id={id_}: {e}")
                time.sleep(SLEEP_BETWEEN)
                continue
        return results
    except Exception as e:
        logging.warning(f"HN fetch error: {e}")
        return []

# ===== Reddit (r/all top of day, public JSON) =====
def fetch_reddit(limit: int = REDDIT_LIMIT) -> List[str]:
    url = f"https://www.reddit.com/r/all/top.json?limit={limit}&t=day"
    try:
        r = requests.get(url, headers=HEADERS, timeout=TIMEOUT)
        r.raise_for_status()
        j = r.json()
        out = []
        for child in j.get("data", {}).get("children", []):
            d = child.get("data", {})
            title = d.get("title") or ""
            title = clean_text(title)
            if title:
                out.append(title)
            if len(out) >= TOP_N:
                break
        return out
    except Exception as e:
        logging.warning(f"Reddit fetch error: {e}")
        return []

# ===== (اختياري) مصادر إضافية يمكن إضافتها لاحقًا =====
# def fetch_other_source(): ...
# ================================================

def combine_and_write(sources: List[List[str]], out_path: Path, top_n: int = TOP_N) -> List[str]:
    seen: Set[str] = set()
    out: List[str] = []
    for src in sources:
        for t in src:
            key = t.lower().strip()
            if not key:
                continue
            if key in seen:
                continue
            seen.add(key)
            out.append(t)
            if len(out) >= top_n:
                break
        if len(out) >= top_n:
            break
    if not out:
        logging.info("No live trends fetched; using fallback sample list.")
        out = SAMPLE_FALLBACK.copy()
    # write file
    try:
        out_path.write_text("\n".join(out), encoding="utf-8")
        logging.info(f"Wrote {len(out)} trends to {out_path.resolve()}")
    except Exception as e:
        logging.error(f"Failed writing trends file: {e}")
    return out

def main():
    logging.info("Starting trends fetch (HN + Reddit) ...")
    # Fetch sources
    hn = fetch_hackernews()
    rd = fetch_reddit()
    combined = []
    # prefer HN first then Reddit, but include both
    if hn:
        combined.append(hn)
    if rd:
        combined.append(rd)
    saved = combine_and_write(combined, OUT_FILE, TOP_N)
    # print numbered list
    for i, t in enumerate(saved, 1):
        logging.info(f"{i}. {t}")
    logging.info("Done.")
    return 0

if __name__ == "__main__":
    try:
        sys.exit(main())
    except KeyboardInterrupt:
        logging.info("Interrupted by user.")
        sys.exit(1)
```
