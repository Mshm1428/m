```python
import os, re, sys, time, json, requests
from pathlib import Path
from hashlib import md5
from difflib import SequenceMatcher

# ========= إعدادات =========
ROOT = Path.cwd()
SRC = ROOT / "publish_ready_cleaned"
PIN_DIR = ROOT / "pin_images"
PUBLISHED = ROOT / "published"
PUBLISHED.mkdir(exist_ok=True)

MIN_WORDS = 1200
BATCH_SIZE = 5   # عدد المقالات في كل تشغيل
SIMILARITY_THRESHOLD = 0.85

BLOG_ID = "1967615204499448730"
TOKEN_URL = "https://oauth2.googleapis.com/token"
BLOG_POSTS_URL = f"https://www.googleapis.com/blogger/v3/blogs/{BLOG_ID}/posts/"

# ========= دوال مساعدة =========

def safe_name(s):
    s = s.strip()
    return "".join(ch if ch.isalnum() or ch in "-_" else "_" for ch in s)[:90]

def word_count(html):
    plain = re.sub(r"<[^>]+>", " ", html)
    return len([w for w in plain.split() if w.strip()])

def similar(a, b):
    return SequenceMatcher(None, a, b).ratio()

def load_oauth():
    # أولوية لمتغيرات البيئة
    cid = os.getenv("GOOGLE_CLIENT_ID")
    csec = os.getenv("GOOGLE_CLIENT_SECRET")
    rtoken = os.getenv("GOOGLE_REFRESH_TOKEN")
    if cid and csec and rtoken:
        return cid, csec, rtoken

    # أو token.json
    token_file = ROOT / "token.json"
    if token_file.exists():
        data = json.loads(token_file.read_text(encoding="utf-8"))
        return data.get("client_id"), data.get("client_secret"), data.get("refresh_token")

    return None, None, None

def exchange_token(cid, csec, rtoken):
    data = {
        "client_id": cid,
        "client_secret": csec,
        "refresh_token": rtoken,
        "grant_type": "refresh_token"
    }
    r = requests.post(TOKEN_URL, data=data, timeout=15)
    r.raise_for_status()
    return r.json()["access_token"]

def image_exists(name):
    p = PIN_DIR / name
    return p.exists() and p.stat().st_size > 0

# ========= النشر =========

def main():
    if not SRC.exists():
        print("❌ مجلد المقالات غير موجود")
        sys.exit(1)

    files = sorted(SRC.glob("*.html"))
    if not files:
        print("ℹ️ لا يوجد مقالات للنشر")
        sys.exit(0)

    cid, csec, rtoken = load_oauth()
    if not (cid and csec and rtoken):
        print("❌ بيانات OAuth غير مكتملة")
        sys.exit(2)

    try:
        access_token = exchange_token(cid, csec, rtoken)
    except Exception as e:
        print("❌ فشل الحصول على access token:", e)
        sys.exit(2)

    headers = {
        "Authorization": f"Bearer {access_token}",
        "Content-Type": "application/json"
    }

    published_count = 0
    published_hashes = set()

    for f in files:
        if published_count >= BATCH_SIZE:
            break

        html = f.read_text(encoding="utf-8")

        wc = word_count(html)
        if wc <= MIN_WORDS:
            print("⏭️ تخطي (قصير):", f.name, wc)
            continue

        h = md5(html.encode("utf-8")).hexdigest()
        if h in published_hashes:
            print("⏭️ تخطي (مكرر):", f.name)
            continue

        title_m = re.search(r"<h1[^>]*>(.*?)</h1>", html, re.I | re.S)
        title = title_m.group(1).strip() if title_m else f.stem

        img_name = safe_name(title) + ".png"
        if not image_exists(img_name):
            print("⏭️ تخطي (لا صورة):", f.name)
            continue

        payload = {
            "kind": "blogger#post",
            "blog": {"id": BLOG_ID},
            "title": title,
            "content": html
        }

        try:
            r = requests.post(BLOG_POSTS_URL, headers=headers, json=payload, timeout=30)
            r.raise_for_status()
            url = r.json().get("url", "")
            print("✅ نُشر:", f.name, "->", url)

            published_hashes.add(h)
            (PUBLISHED / f.name).write_text(html, encoding="utf-8")
            f.unlink()

            published_count += 1
            time.sleep(2)

        except Exception as e:
            print("❌ فشل النشر:", f.name, e)

    print("🎉 انتهى — عدد المقالات المنشورة:", published_count)

if __name__ == "__main__":
    main()
```
